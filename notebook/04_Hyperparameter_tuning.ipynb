```python
import pandas as pd
import numpy as np
from sklearn.ensemble import RandomForestRegressor,GradientBoostingRegressor 
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
X_train = pd.read_csv(r"C:\Users\mayak\OneDrive\Desktop\fastapi\X_train_processed.xls")
X_test = pd.read_csv(r"C:\Users\mayak\OneDrive\Desktop\fastapi\X_test_processed.xls")

y_train = pd.read_csv(r"C:\Users\mayak\OneDrive\Desktop\fastapi\y_train.xls").squeeze()
y_test = pd.read_csv(r"C:\Users\mayak\OneDrive\Desktop\fastapi\y_test.xls").squeeze()
```

```python
rf_model = RandomForestRegressor(random_state=98)
```

```python
rf_model.fit(X_train,y_train)
```

```python
rf_pred = rf_model.predict(X_test)

rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("Tuned Random Forest Results")
print("---------------------------")
print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R2  :", rf_r2)
```

```python
rf_params = {
    'n_estimators': [100, 200],
    'max_depth': [None, 10, 20],
    'min_samples_split': [2, 5]
}
```

```python
rf_grid = GridSearchCV(
    estimator=rf_model,
    param_grid=rf_params,
    cv=3,
    scoring='r2',
    n_jobs=-1
)
```

```python
rf_grid.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(rf_grid.best_params_)
```

```python
best_rf = rf_grid.best_estimator_ 
```

```python
best_rf = rf_grid.best_estimator_

rf_tuned_pred = best_rf.predict(X_test)

rf_tuned_mae = mean_absolute_error(y_test, rf_tuned_pred)
rf_tuned_rmse = np.sqrt(mean_squared_error(y_test, rf_tuned_pred))
rf_tuned_r2 = r2_score(y_test, rf_tuned_pred)

print("Tuned Random Forest Results")
print("---------------------------")
print("MAE :", rf_tuned_mae)
print("RMSE:", rf_tuned_rmse)
print("R2  :", rf_tuned_r2)
```

```python
print("Random Forest Baseline")
print("----------------------")
print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R²  :", rf_r2)

print("\nTuned Random Forest")
print("-------------------")
print("MAE :", rf_tuned_mae)
print("RMSE:", rf_tuned_rmse)
print("R2  :", rf_tuned_r2)
```

```python
gb_model = GradientBoostingRegressor(
    random_state=42
)
```

```python
gb_model.fit(X_train,y_train)
```

```python
gb_pred = gb_model.predict(X_test)

gb_mae = mean_absolute_error(y_test, gb_pred)
gb_rmse = np.sqrt(mean_squared_error(y_test, gb_pred))
gb_r2 = r2_score(y_test, gb_pred)

print("base Gradient boosting Results")
print("---------------------------")
print("MAE :", gb_mae)
print("RMSE:", gb_rmse)
print("R2  :", gb_r2)
```

```python
gb_params = {
    'n_estimators': [100, 200],
    'learning_rate': [0.05, 0.1],
    'max_depth': [2, 3],
    'min_samples_split': [2, 5]
}
```

```python
gb_grid = GridSearchCV(
    estimator=gb_model,
    param_grid=gb_params,
    cv=4,
    scoring='r2',
    n_jobs=-1
)
```

```python
gb_grid.fit(X_train, y_train)
```

```python
print("Best Parameters:")
print(gb_grid.best_params_)
```

```python
print("Best CV R2:")
print(gb_grid.best_score_)
```

```python
best_gb = gb_grid.best_estimator_

gb_tuned_pred = best_gb.predict(X_test)

gb_tuned_mae = mean_absolute_error(y_test, gb_tuned_pred)
gb_tuned_rmse = np.sqrt(mean_squared_error(y_test, gb_tuned_pred))
gb_tuned_r2 = r2_score(y_test, gb_tuned_pred)

print("Tuned Gradient Boosting Results")
print("--------------------------------")
print("MAE :", gb_tuned_mae)
print("RMSE:", gb_tuned_rmse)
print("R²  :", gb_tuned_r2)
```

```python
tuned_results = pd.DataFrame({
    'Model': [
        'Tuned Random Forest',
        'Tuned Gradient Boosting'
    ],
    'MAE': [
        rf_tuned_mae,
        gb_tuned_mae
    ],
    'RMSE': [
        rf_tuned_rmse,
        gb_tuned_rmse
    ],
    'R2': [
        rf_tuned_r2,
        gb_tuned_r2
    ]
})

tuned_results
```
