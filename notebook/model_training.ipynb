```python
import pandas as pd
import numpy as np

from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import RandomForestRegressor
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error
from sklearn.metrics import r2_score
!pip install xgboost
from xgboost import XGBRegressor
```

```python
X_train = pd.read_csv(r"C:\Users\mayak\OneDrive\Desktop\fastapi\X_train_processed.xls")
X_test = pd.read_csv(r"C:\Users\mayak\OneDrive\Desktop\fastapi\X_test_processed.xls")

y_train = pd.read_csv(r"C:\Users\mayak\OneDrive\Desktop\fastapi\y_train.xls").squeeze()
y_test = pd.read_csv(r"C:\Users\mayak\OneDrive\Desktop\fastapi\y_test.xls").squeeze()
```

```python
print("X_train:", X_train.shape)
print("X_test :", X_test.shape)
print("y_train:", y_train.shape)
print("y_test :", y_test.shape)
```

```python
model = LinearRegression()
```

```python
model.fit(X_train, y_train)
```

```python
y_pred_logistic=model.predict(X_test)
```

```python
print(y_pred_logistic[:10])
```

```python
mae = mean_absolute_error(y_test,y_pred_logistic)

print("MAE:", mae)
```

```python
rmse = np.sqrt(mean_squared_error(y_test, y_pred_logistic))

print("RMSE:", rmse)
```

```python
r2 = r2_score(y_test, y_pred_logistic)

print("R² Score:", r2)
```

```python
print("Linear Regression Results")
print("-------------------------")
print("MAE :", mae)
print("RMSE:", rmse)
print("R²  :", r2)
```

```python
dt_model = DecisionTreeRegressor(random_state=42)
```

```python
dt_model.fit(X_train, y_train)
```

```python
dt_pred = dt_model.predict(X_test)
```

```python
dt_mae = mean_absolute_error(y_test, dt_pred)
dt_rmse = np.sqrt(mean_squared_error(y_test, dt_pred))
dt_r2 = r2_score(y_test, dt_pred)

print("Decision Tree Results")
print("---------------------")
print("MAE :", dt_mae)
print("RMSE:", dt_rmse)
print("R²  :", dt_r2)
```

```python
rf_model = RandomForestRegressor(random_state=52)
```

```python
rf_model.fit(X_train, y_train)
```

```python
rf_pred = rf_model.predict(X_test)
```

```python
rf_mae = mean_absolute_error(y_test, rf_pred)
rf_rmse = np.sqrt(mean_squared_error(y_test, rf_pred))
rf_r2 = r2_score(y_test, rf_pred)

print("Random Forest Results")
print("---------------------")
print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R²  :", rf_r2)
```

```python
gb_model = GradientBoostingRegressor(random_state=54)
```

```python
gb_model.fit(X_train, y_train)
```

```python
gb_pred = gb_model.predict(X_test)
```

```python
gb_mae = mean_absolute_error(y_test, gb_pred)
gb_rmse = np.sqrt(mean_squared_error(y_test, gb_pred))
gb_r2 = r2_score(y_test, gb_pred)

print("Gradient Boosting Results")
print("-------------------------")
print("MAE :", gb_mae)
print("RMSE:", gb_rmse)
print("R²  :", gb_r2)
```

```python
xgb_model = XGBRegressor(random_state=28)
```

```python
xgb_model.fit(X_train, y_train)
```

```python
xgb_pred = xgb_model.predict(X_test)
```

```python
xgb_mae = mean_absolute_error(y_test, xgb_pred)
xgb_rmse = np.sqrt(mean_squared_error(y_test, xgb_pred))
xgb_r2 = r2_score(y_test, xgb_pred)

print("XGBoost Results")
print("---------------")
print("MAE :", xgb_mae)
print("RMSE:", xgb_rmse)
print("R²  :", xgb_r2)
```

```python
print("Model Performance")
print("------------------")

print("Linear Regression R² :", r2)
print("Decision Tree R²     :", dt_r2)
print("Random Forest R²     :", rf_r2)
print("Gradient Boosting R² :", gb_r2)
print("XGBoost R²           :", xgb_r2)
```
