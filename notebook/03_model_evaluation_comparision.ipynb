```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
results = pd.DataFrame({
    'Model': [
        'Linear Regression',
        'Decision Tree',
        'Random Forest',
        'Gradient Boosting',
        'XGBoost'
    ],
    'MAE': [
        44.749212397379765,
        5.068060,
        3.665561,
        6.508809,
        5.903302
    ],
    'RMSE': [
        121.45696030097047,
        38.147875,
        37.401150,
        32.013393,
        41.535217
    ],
    'R2': [
        0.49246825827197294,
        0.949932,
        0.951873,
        0.964740,
        0.940646
    ]
})

results
```

```python
plt.figure(figsize=(10, 5))

plt.bar(results['Model'], results['R2'])

plt.xlabel('Model')
plt.ylabel('R² Score')
plt.title('Model Comparison - R² Score')

plt.xticks(rotation=20)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(results['Model'], results['MAE'])

plt.xlabel('Model')
plt.ylabel('MAE')
plt.title('Model Comparison - MAE')

plt.xticks(rotation=20)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.bar(results['Model'], results['RMSE'])

plt.xlabel('Model')
plt.ylabel('RMSE')
plt.title('Model Comparison - RMSE')

plt.xticks(rotation=20)
plt.tight_layout()

plt.show()
```

```python
results.to_csv(
    "model_comparison.csv",
    index=False
)
```
